```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

flights_data = pd.read_csv('flights.csv')
flights_data.head(10)
weather_data_pd = pd.read_csv('weather.csv')
weather_data_np = weather_data_pd.to_numpy()

# Question 1
q_1 = len(flights_data[(flights_data['origin'] == 'JFK') & (flights_data['dest'] == 'SLC')])
print("q_1:", q_1)

# Question 2
q_2 = flights_data[flights_data['dest'] == 'SLC']['carrier'].nunique()
print("q_2:", q_2)   # 2

# Question 3
q_3 = flights_data[flights_data['dest'] == 'RDU']['arr_delay'].mean()
print("q_3:", q_3)

# Question 4
sea_flights = flights_data[flights_data['dest'] == 'SEA']
q_4 = sea_flights['origin'].isin(['LGA', 'JFK']).mean()
print("q_4:", q_4)

# Question 5
# Build a date column in y/m/d format (2013/1/1)
flights_data['date'] = (flights_data['year'].astype(str) + '/' +
                        flights_data['month'].astype(str) + '/' +
                        flights_data['day'].astype(str))
q_5 = flights_data.groupby('date')['dep_delay'].mean().nlargest(1)
print("q_5:\n", q_5)

# Question 6
q_6 = flights_data.groupby('date')['arr_delay'].mean().nlargest(1)
print("q_6:\n", q_6)

# Question 7
# speed = distance / air_time

fast = flights_data[flights_data['origin'].isin(['LGA', 'JFK']) &
                    (flights_data['year'] == 2013)].copy()
fast['speed'] = fast['distance'] / fast['air_time']
q_7 = fast.nlargest(1, 'speed')[['tailnum', 'speed']]
print("q_7:\n", q_7)

# Question 8
q_8 = weather_data_pd.fillna(0)
print("q_8 NaNs left:", q_8.isna().sum().sum())
# Use weather_data_np
# Column positions: month = 3, humid = 8 (column 0 is the unnamed index column)

# Question 9
feb = weather_data_np[weather_data_np[:, 3] == 2]
print("q_9:", q_9)

# Question 10
feb_humid = feb[:, 8].astype(float)
q_10 = feb_humid.mean()
print("q_10:", q_10)

# Question 11
q_11 = feb_humid.std()
print("q_11:", q_11)
```
